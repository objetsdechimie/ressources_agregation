```python
from graphviz import Digraph

def generer_arbre_extraction_joli(K, nb_plateaux, V_orga, V_aq):
    """
    Génère un bel arbre d'extraction liquide-liquide successive.
    """
    # Calcul des fractions théoriques par étage
    q = 1 / (1 + K * (V_orga / V_aq))  # Reste en phase aqueuse
    p = 1 - q                          # Passe en phase organique
    
    # Configuration du graphique avec un style épuré
    dot = Digraph(comment='Arbre Extraction Liquide-Liquide', format='png')
    dot.attr(rankdir='TB', size='12,12', splines='ortho')
    
    # Styles globaux par défaut pour éviter les cases noires et les traits lourds
    dot.attr('node', fontname='Helvetica', fontsize='11', shape='box', style='filled,rounded', penwidth='1.5')
    dot.attr('edge', fontname='Helvetica', fontsize='10', color='#7f8c8d', penwidth='1.2')
    
    # Nœud initial (Départ)
    dot.node('A0', f'DÉPART\nPhase Aqueuse : 100%\n(V = {V_aq} mL)', 
             fillcolor='#d6e4f0', color='#34495e', fontcolor='#2c3e50')
    
    id_aq_actuel = 'A0'
    fraction_restante_aq = 1.0
    
    for i in range(1, nb_plateaux + 1):
        # Fractions absolues récupérées à cet étage
        frac_orga_etage = fraction_restante_aq * p
        frac_aq_etage = fraction_restante_aq * q
        
        id_orga = f'O{i}'
        id_aq_suivant = f'A{i}'
        
        # Phase Organique : Vert pastel doux (#e8f5e9) avec bordure verte (#2ecc71)
        dot.node(id_orga, f'Étage {i} : Phase Organique\nExtrait : {frac_orga_etage:.2%}\n(V = {V_orga} mL)', 
                 fillcolor='#e8f5e9', color='#2ecc71', fontcolor='#27ae60')
        
        # Phase Aqueuse : Bleu pastel doux (#e3f2fd) avec bordure bleue (#3498db)
        dot.node(id_aq_suivant, f'Étage {i} : Phase Aqueuse\nRestant : {frac_aq_etage:.2%}\n(V = {V_aq} mL)', 
                 fillcolor='#e3f2fd', color='#3498db', fontcolor='#2980b9')
        
        # Flèches directionnelles
        dot.edge(id_aq_actuel, id_orga, label=f' Vers Orga ({p:.1%})')
        dot.edge(id_aq_actuel, id_aq_suivant, label=f' Reste Eau ({q:.1%})')
        
        id_aq_actuel = id_aq_suivant
        fraction_restante_aq = frac_aq_etage
        
    # Calcul du bilan
    total_recupere = 1.0 - fraction_restante_aq
    
    # Bilan final : Jaune/Or doux (#fff9c4)
    dot.node('Bilan', f'BILAN GLOBAL\nTotal Extrait (Orga) : {total_recupere:.2%}\nTotal Perdu (Eau) : {fraction_restante_aq:.2%}', 
             shape='note', fillcolor='#fff9c4', color='#f1c40f', fontcolor='#7f8c8d')
    dot.edge(id_aq_actuel, 'Bilan', style='dashed', constraint='true')

    # Rendu
    nom_fichier = 'arbre_extraction_propre'
    dot.render(nom_fichier, view=True)
    print(f"L'arbre a été généré avec succès dans '{nom_fichier}.png' !")

# --- Vos paramètres ---
K_choisi = 18.0          
plateaux = 4           
V_organique = 100        
V_aqueux = 100          

generer_arbre_extraction_joli(K=K_choisi, nb_plateaux=plateaux, V_orga=V_organique, V_aq=V_aqueux)
```
